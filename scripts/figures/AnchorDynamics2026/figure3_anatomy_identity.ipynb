# Figure 3 — where MEC spatial cells sit, and what position predicts

Figure 2 shows that grid cells and putative interneurons follow the population
anchoring state. This asks whether that has an anatomical address.

**One axis does everything.** Mediolateral position predicts both what a cell is
and whether it follows; dorsoventral position, anterior-posterior position and
probe depth predict neither; layer predicts identity only.

| axis | predicts | within session |
|---|---|---|
| **M–L** | grid identity | ρ = −0.078, 18/27 sessions, **p = 0.025** |
| **M–L** | following | ρ = −0.109, 19/27 sessions, **p = 0.034** |
| **layer** | grid identity | 8.9% sup vs 3.9% deep, 21/31, **p = 0.002** |
| D–V | grid identity | p = 0.90 |
| D–V | following | p = 0.99 |
| A–P | grid identity | p = 0.10 |
| depth | following | p = 0.92 |

## Which sessions can answer the mediolateral question

Mediolateral position is assigned **per shank**, so a single-shank session has
no mediolateral variation at all and contributes only noise — 33 of 61 sessions
span under 50 µm. The within-session test runs on the **27 sessions spanning at
least 200 µm**, two shank pitches or more.

This matters more than it sounds. An earlier version selected sessions by
`nunique > 3`, the count of distinct mediolateral values. That rule *excludes*
the cleanly sampled 2-, 3- and 4-shank sessions — a 3-shank session holds three
discrete values while spanning 510 µm — and *admits* single-shank sessions whose
coordinates differ only by sub-micron numerical jitter. It ran the test almost
entirely on sessions with no mediolateral variation, and reported **p = 1.00**
for an effect that is present at **p = 0.034**.

**The control that makes the restriction honest:** for dorsoventral position and
probe depth the criterion changes nothing (p = 0.99 and 0.92 either way),
because every session varies along the shank. Only the axis sampled by shanks is
affected, so the restriction is not manufacturing the effect.

## What cannot be settled here

Mediolateral position predicts identity, and identity predicts following, so the
mediolateral gradient in following may be nothing more than the identity
gradient seen through it. Restricting to non-grid cells leaves the effect size
almost unchanged (ρ = −0.100 against −0.109) but loses significance at 26
sessions (p = 0.111). The two accounts are not separable at this sample size and
the figure does not claim otherwise.

## Panels

- **A** the gradient shown rather than summarised: open-field rate maps from four
  M25 sessions at different mediolateral positions (≈3000 → ≈3720 µm), sampled
  **proportionally** so the grid:non-grid ratio drawn is the ratio actually
  recorded. Red borders are grid cells, blue non-grid.
- **B, C** best-fit slices (M25, all mice) under small 3D views of the probe in
  MEC. Both axes are micrometres and the panels are drawn to **equal aspect**,
  so the slice looks like the tissue rather than being stretched to fill its cell.
- **D–F** identity across the axes: grid fraction against mediolateral and
  dorsoventral position, then superficial vs deep paired within session.
  Anterior-posterior is shown as layer rather than as a coordinate, because in a
  structure this curved the AP coordinate is largely a proxy for layer, and layer
  is the quantity with a mechanism attached.
- **G–I** the same axes against **following**, pooled and within session.
- **J** the same question between **regions**, scored against each cell's own
  circular-shift null. This is necessary rather than cosmetic: chance differs by
  region (+0.130 to +0.157), so ranking regions on raw correlation partly ranks
  them on how often their cells happen to be anchored. Against the null, only
  the entorhinal groups exceed chance and **visual cortex sits below its own
  null**.
