# Figure 4 — the local circuit: wired together, but the wiring does not carry the state

Figure 2 shows that grid cells and putative interneurons follow the population
anchoring state, and follow **each other**. That is a trial-level statistic, and
it invites a circuit explanation: perhaps the cells that share a state are the
cells that are synaptically connected. This figure tests that directly, and the
answer is what pivots the paper toward arousal (Figure 5).

The main figure leans on this group hard: interneurons are one of the two
identities that follow the population state (2D), they pair with grid cells
(2G). So the group has to be shown to be real rather than a threshold applied
to a histogram.

**The call (A).** Peak-to-trough does the work — a two-component Gaussian
mixture on log duration, cut at its antimode, components **0.25 vs 0.69 ms
separated by 4.0 SD** — but the boundary applied is a fixed **0.4 ms**, not the
fitted antimode. A refitted cut moves with whichever cells are supplied, so
sessions, mice and reanalyses would each get a slightly different boundary and
cells would change class for reasons unrelated to the cells. The fit is kept as
corroboration and drawn beside it: it lands at 0.42 ms, within a histogram bin
of 0.4, and the empirical density minimum is at 0.38–0.40 ms. The choice moves
12 cells of 842 either way.

Firing rate is *not* fitted that way either: log rate separates at only
**1.8 SD** and has no usable antimode, so the top quartile (10.2 Hz) is a second
requirement laid on the waveform call. It is conservative — it rejects **43% of
narrow-spiking cells** (610 of 1,440), which fall back to their open-field class.

**The validation is physiological, not statistical (B–E).** The old version of
this supplement validated the group with open-field speed modulation. That was
true but weak: speed modulation is common in MEC and not specific to inhibition,
so the test could not have failed informatively. The monosynaptic test can. The
detector works on spike-train cross-correlograms and has **no access to waveform
shape, firing rate, or the class labels** — so if panel A were an arbitrary cut
there would be no reason for its members to be the cells that suppress their
targets.

They are. Inhibitory connections originate from the cells panel A called
interneurons at **3.1× their share of the population** (D), while excitatory
connections do not, and they arrive with the **slower kinetics** expected of
GABA_A rather than AMPA (median 3.0 vs 1.0 ms, G).

Detecting them required extending the detector. The published criteria find a
causal *peak* and therefore excitation only; an inhibitory connection is a
*trough* and was invisible in principle, which would have made panel E
unanswerable rather than underpowered. Mirroring every criterion onto the lower
Poisson tail gives **1,059 excitatory and 7,775 inhibitory** connections at
matched jitter-null FDRs (0.09 / 0.10, panel H).

**And the connectivity answers a second question for free (F, H).** It shows
grid cells and interneurons are wired together — `int → grid` is the strongest
pathway in the dataset at 3.93%, 5.1× over within-session identity shuffling,
*p* = 0.0005 (E here; F is its excitatory complement, where
`grid → grid` leads at 0.217%, 2.6×, *p* = 0.004 — both matrices are shown
because neither is interpretable alone, the excitatory one being an order of
magnitude sparser throughout). That is what makes the
trial-level pairing in 2G a subnetwork rather than two populations tracking one
signal.

But the state does **not** travel through those synapses (I). Against
unconnected pairs matched for session, identity pair and probe distance
(±40 µm):

| pair | connected | matched | *p* | MDE | verdict |
|---|---|---|---|---|---|
| grid+int | +0.176 (n=193) | +0.187 | 0.39 | 0.045 | **informative null** |
| int+int | +0.090 (n=116) | +0.090 | 0.90 | 0.043 | **informative null** |
| other | +0.068 (n=3238) | +0.062 | 0.032 | 0.009 | +0.006, trivial |
| grid+grid | +0.510 (n=27) | +0.465 | 0.31 | **0.157** | **underpowered** |

**MDE** is the minimum effect detectable at 80% power, and it is what decides
whether a null means anything — the effect to beat is the identity effect of the
main figure, ~0.11. Grid–interneuron would have caught a wiring effect a third
that size and saw nothing. Grid–grid, at 27 connected pairs, could not have
resolved an effect as large as the entire identity effect, so its *p* = 0.31 is
labelled *underpowered* rather than counted as a null.

Anchoring travels with what a cell **is**, not with what it is wired to.
